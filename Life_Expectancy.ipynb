# Missing value Tereatment

```python
df.isnull().sum()
```

```python
for col in [" BMI ","Polio","Income composition of resources"]:
  df[col].fillna(df[col].median(),inplace=True)
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import KNNImputer
```

```python
kni = KNNImputer()
```

```python
for i in df.select_dtypes(include='number').columns:
  df[i]=kni.fit_transform(df[[i]])
```

```python
df.isnull().sum()
```

```python
def wisker(col):
  q1,q3 = np.percentile(col,[25,27])
  iqr = q3-q1
  upper_bound = q3+1.5*iqr
  lower_bound = q1-1.5*iqr

  return lower_bound,upper_bound
```

```python
df
```

```python
df.columns
```

```python
df['GDP'].max()
```

```python
for i in ['GDP','Total expenditure',' thinness  1-19 years',' thinness 5-9 years']:
  lower_bound,upper_bound = wisker(df[i])
  df[i] = np.where(df[i]<lower_bound,lower_bound,df[i])
  df[i] = np.where(df[i]>upper_bound,upper_bound,df[i])
```
